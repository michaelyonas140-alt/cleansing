### 1. Memuat Data (Loading Data)


```python
import pandas as pd
import numpy as np

# Memuat dataset
file_path = '/content/creditcard.csv'
try:
    df = pd.read_csv(file_path)
    print("Data berhasil dimuat!")
    display(df.head())
except Exception as e:
    print(f"Gagal memuat data: {e}")
```

### 2. Memeriksa Informasi Data dan Nilai yang Hilang (Missing Values)

```python
# Menampilkan informasi ringkas tentang dataset
print("=== Informasi Dataset ===")
df.info()

# Memeriksa jumlah nilai null/missing value di setiap kolom
missing_values = df.isnull().sum()
print("\n=== Nilai yang Hilang per Kolom ===")
print(missing_values[missing_values > 0] if missing_values.sum() > 0 else "Tidak ada nilai yang hilang!")
```

### 3. Memeriksa dan Menghapus Duplikasi Data

```python
# Memeriksa jumlah data duplikat
duplicate_count = df.duplicated().sum()
print(f"Jumlah baris duplikat: {duplicate_count}")

if duplicate_count > 0:
    # Menghapus duplikat dan mempertahankan baris pertama
    df = df.drop_duplicates(keep='first')
    print("Data duplikat berhasil dihapus.")
    print(f"Ukuran dataset sekarang: {df.shape}")
```

### 4. Menyimpan Data yang Telah Bersih

```python
# Menghapus baris yang memiliki missing value (hanya 1 baris)
df_clean = df.dropna()
print(f"Ukuran dataset setelah menghapus baris kosong: {df_clean.shape}")

# Menyimpan hasil cleansing ke file baru
clean_file_path = '/content/creditcard_clean.csv'
df_clean.to_csv(clean_file_path, index=False)
print(f"Data bersih final berhasil disimpan di: {clean_file_path}")
```
